```python
df.rename(columns={
    'oldbalanceOrg': 'oldbalanceSender',
    'newbalanceOrig': 'newbalanceSender',
    'oldbalanceDest': 'oldbalanceReceiver',
    'newbalanceDest': 'newbalanceReceiver'
}, inplace=True)
```

```python
 #Flag if sender ends up with a negative balance
df['flag_sender_negative_balance'] = (df['newbalanceSender'] < 0).astype(int)
```

```python

#Flag if receiver ends up with a negative balance
df['flag_receiver_negative_balance'] = (df['newbalanceReceiver'] < 0).astype(int)
```

```python
df.head()
```

```python
#senderside flag
df['flag_senderside'] = ((df['oldbalanceSender'].round(2) - df['newbalanceSender'].round(2) - df['amount']).abs() > 1e-6).astype(int)
```

```python
df.head()
```

```python
#recieverside flag
valid_receiver_txn = df['type'].isin(['TRANSFER', 'CASH_OUT'])

receiver_diff = (df['newbalanceReceiver'].round(2) - df['oldbalanceReceiver'].round(2) - df['amount'].round(2)).abs()

receiver_skipped = (df['oldbalanceReceiver'].round(2) == 0) & (df['newbalanceReceiver'].round(2) == 0)

df['flag_receiverside'] = ((receiver_diff > 1e-6) & ~receiver_skipped & valid_receiver_txn).astype(int)


```

now moving on to phantom transactions the untracked ones



```python
#Flag phantom transactions
df['flag_disappeared'] = (
    df['type'].isin([1, 4]) &  # TRANSFER or CASH_OUT (numeric)
    (df['flag_senderside'] == 0) &  # Sender deducted correctly
    ((df['oldbalanceReceiver'] - df['newbalanceReceiver']).abs() < 1e-6) &  # Receiver saw no change
    (df['amount'] > 0)  # Actual money moved
).astype(int)
```

```python
#Group by sender and compute median transaction amount
sender_median = df.groupby('nameOrig')['amount'].median().rename('median_transactionamount')
#Merge back into the main DataFrame
df = df.merge(sender_median, left_on='nameOrig', right_index=True)
```

```python
df.head()
```

```python
#Flag transactions that are 3 times or more above personal median
df['flag_above_median_spend'] = (
    df['type'].isin([1, 4]) &  #Only apply to TRANSFER and CASH_OUT
    (df['amount'] > 3 * df['median_transactionamount'])
).astype(int)
```

```python
df.head()
```

```python
#flag transactions where amount is zero but balances change
#This checks if the transaction type is TRANSFER or CASH_OUT, amount is zero, and
#balances change for either sender or receiver, excluding cases where all balances are zero

df['flag_amount_zero_but_balances_change'] = (
    (df['type'].isin([1, 4])) &
    (df['amount'].round(2) == 0) &
    (
        (df['oldbalanceSender'].round(2) != df['newbalanceSender'].round(2)) |
        (df['oldbalanceReceiver'].round(2) != df['newbalanceReceiver'].round(2))
    ) &
    ~(
        (df['oldbalanceSender'].round(2) == 0) &
        (df['oldbalanceReceiver'].round(2) == 0) &
        (df['newbalanceSender'].round(2) == 0) &
        (df['newbalanceReceiver'].round(2) == 0)
    )
).astype(int)
```

```python
#Sort data for sequence-based logic
df_sorted = df.sort_values(by=['nameDest', 'step']).reset_index(drop=True)

#Shift relevant columns to get the *next transaction* by the same account
df_sorted['next_step'] = df_sorted['step'].shift(-1)
df_sorted['next_type'] = df_sorted['type'].shift(-1)
df_sorted['next_amount'] = df_sorted['amount'].shift(-1)
df_sorted['next_nameOrig'] = df_sorted['nameOrig'].shift(-1)

#Build the fast cashout flag
df_sorted['flag_fast_cashout'] = (
    (df_sorted['type'] == 1) &  # TRANSFER
    (df_sorted['next_type'] == 4) &  # followed by CASH_OUT
    (df_sorted['nameDest'] == df_sorted['next_nameOrig']) &  # same person
    (df_sorted['amount'].round(2) == df_sorted['next_amount'].round(2)) &  # same amount
    ((df_sorted['next_step'] - df_sorted['step']) <= 2)  # fast: within 2 time steps
).astype(int)

df['flag_fast_cashout'] = df_sorted['flag_fast_cashout']
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import f1_score, make_scorer
import pandas as pd
import numpy as np

# Prepare data
df_sample = df.sample(n=200000, random_state=42)
columns_to_drop = [col for col in ['isFraud', 'nameOrig', 'nameDest'] if col in df.columns]
X = df_sample.drop(columns=columns_to_drop)
y_true = df['isFraud']

# Custom scoring function using F1 Score
def f1_scorer(estimator, X):
    y_pred = estimator.predict(X)
    y_pred = pd.Series(y_pred).map({1: 0, -1: 1})
    return f1_score(y_true, y_pred)

# Define pipeline
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('isoforest', IsolationForest(random_state=42))
])

# Define parameter distribution
param_dist = {
    'isoforest__n_estimators': [100, 200, 300],
    'isoforest__max_samples': ['auto', 0.5, 0.7, 0.9],
    'isoforest__contamination': np.linspace(0.001, 0.01, 10),
    'isoforest__max_features': [0.6, 0.8, 1.0]
}

# Setup RandomizedSearchCV
random_search = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_dist,
    n_iter=20,
    scoring=f1_scorer,
    cv=2,
    verbose=2,
    random_state=42,
    n_jobs=-1
)

# Run the search
random_search.fit(X)

# Best model and parameters
best_model = random_search.best_estimator_
print("Best Parameters:", random_search.best_params_)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score, precision_score, recall_score, f1_score
import numpy as np
import pandas as pd

# Rebuild pipeline with best parameters
final_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('isoforest', IsolationForest(
        n_estimators=300,
        max_samples=0.7,
        max_features=0.6,
        contamination=0.007,
        random_state=42
    ))
])

# Fit the pipeline
final_pipeline.fit(X)

# Predict and map output
y_true = df_sample['isFraud'].reset_index(drop=True)
y_pred = final_pipeline.predict(X)
y_pred = pd.Series(y_pred).map({1: 0, -1: 1})  # 0: normal, 1: anomaly/fraud

# Evaluation
print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification Report:\n", classification_report(y_true, y_pred))
print("ROC-AUC Score:", roc_auc_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred))
print("Recall:", recall_score(y_true, y_pred))
print("F1 Score:", f1_score(y_true, y_pred))

# Optional: Add predictions to your sampled dataframe
df_sample['anomaly_score'] = y_pred


accuracy = accuracy_score(y_true, y_pred)
print("Accuracy:", accuracy)
```
